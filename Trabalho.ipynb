```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Gerando a Normal Zk

def Normal():
    u = np.random.uniform()
    u_linha = 2*np.pi*u
    t = -np.log(u)
    x = np.sqrt(2*t)*u
```

```python
t = .5
n = 1000
dt = t/n
time = np.linspace(0,t,n)

simulacoes = 10
for i in range(simulacoes):
    x = 0
    trajetoria = []

    for k in range(n):
        x = x + np.sqrt(dt)*Normal()
        trajetoria.append(x)

    plt.plot(time, trajetoria, linewidth = 1)


plt.xlabel('Tempo')
plt.ylabel('Trajetória')
plt.title('Trajetória do Movimento Browniano')
plt.show()
```
