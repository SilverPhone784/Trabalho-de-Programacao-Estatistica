```python
#Gerando binomial via recursão
def bin_neg_rec(r, p, k):
  amostrinha = []
  for _ in range(k):
    u = np.random.uniform()
    n = r
    pi = p**r
    F = pi
    while u > F:
      pi = pi*n*(1-p)/(n-r+1)
      n += 1
      F += pi
    amostrinha.append(n-r)
  return amostrinha
```

```python
def f(x, taxa, alpha):
  return (taxa**alpha)*np.exp(-x*taxa)*(x**(alpha-1))/math.gamma(alpha)
```

```python
def g(x, mi):
  return np.exp(-mi*x)*mi
```
